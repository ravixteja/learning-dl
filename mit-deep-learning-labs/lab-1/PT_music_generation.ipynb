## 2.3 Process the dataset for learning
Well our main objective here is prediction. We are trying to train an RNN model to learn patterns in ABC music, and then use this model to generate(in the sense, predict) a new piece of music based on learned info.
<br><br>
In short, given a sequence of characters to the model, what is the most probable next character?
<br><br>
To achieve this, we will input sequence of characters to the model, train it to predict the output (the most probable next character at each time step)
<br><br>
RNNs ability to maintain an internal state which depends on previously seen inputs, thus able to maintain information about characters seen till a given moment will help us in generating the outcome(prediction).
<br><br>

### Vectorize
From the dataset, we have extracted unique characters -- we call this as corpus / dictionary. We will now convert these characters into numerical representations (vectorise).
<br><br>
We will now generate two lookup tables, one mapping characters to numbers and vice versa.

```python
display(vocab)
```

```python
# Define numerical representation

# lookup table - maps character to numbers (here, index)
char2idx = {char : i for i, char in enumerate(vocab)}

# lookup table - maps numbers to characters (here, index)
idx2char = np.array(vocab)
```

This generates numerical representation of the corpus. Unique characters in `vocab` are mapped to indices from 0 to `len(vocab)`

```python
# Taking a look into how these characters and indices are represented
for char, i in char2idx.items():
    print(f"{repr(char)} : {i}")
```

```python
# Now vectorize songs string

def vectorize_string(string):

    # split string to characters
    char_list = [character for character in string]
    # print(char_list)
    vectorized = []

    # match characters with char2idx to convert char_list to indices array
    for character in char_list:
        vectorized.append(char2idx[character])
    # print(vectorized)
    # print(len(vectorized))
    return np.array(vectorized)
```

### Create training samples (inputs and targets)
We will now divide text into example sequences(pair of input and an output/target) to be used during training.
<br><br>
Each sequence contains:
- a uniform length of characters acting as input
- a same length of characters (shifted by one character right) acting as target

These pairs will be used to train our RNN to predict the next character.
<br><br>
This is batch method of training

```python
# define a function to create batches / training pairs
def create_batch(vectorized_songs, seq_len, batch_size):
    # get last index of `vectorized_songs`
    n = vectorized_songs.shape[0] - 1

    # randomly choose starting indices for examples
    idx = np.random.choice(n - seq_len, batch_size)

    # construct list of input and output sequences
    # input_batch = []
    # output_batch = []
    # for index in idx:
        # pick_ip = vectorized_songs[index:(index+seq_len)]
        # input_batch.append(pick_ip)
        # pick_op = vectorized_songs[(index+1):(index+seq_len+1)]
        # output_batch.append(pick_op)
    input_batch = [vectorized_songs[index : (index+seq_len)] for index in idx]
    output_batch = [vectorized_songs[(index+1):(index+seq_len+1)] for index in idx]
    
    input_batch = np.array(input_batch)
    output_batch = np.array(output_batch)

    x_batch = torch.tensor(input_batch, dtype = torch.long)
    y_batch = torch.tensor(output_batch, dtype = torch.long)
    # print(x_batch)
    # print(y_batch)
    return x_batch, y_batch
```

## 2.4 The Recurrent Neural Network (RNN) model
Now, it's time to define and train an RNN model using batches of snippets generated in the previous few steps.
<br><br>
The model we will be building is based off the LSTM Architecture - which uses state vector to maintain information about temporal relationships between consecutive characters.
<br><br>
Output of LSTM is fed into fully connected `nn.Linear` layer - which outputs a softmax over each character in the `vocab`, and then sample from this distribution to predict next character.
<br><br>
We use `nn.Module` to define a model with the following three components:
- nn.Embedding : input layer, consisting of trainable lookup table, mapping numbers of each characters to a vector with a specific dimension.
- nn.LSTM : The LSTM network
- nn.Linear : Output layer with size = len(vocab)
<br>
<br>

![alt text](architecture.jpg)

### Now define RNN Model

```python
# Defining the RNN Model
class OurLSTMModel(nn.Module):

    # define constructor
    def __init__(self, embedding_dim, hidden_size, vocab_size):
        # call constructor of nn.Module
        super(OurLSTMModel, self).__init__()
        self.hidden_size = hidden_size

        # define each layer
        # Layer 1 - Embedding Layer - transform indices to dense vectors of fixed size
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        # Layer 2 - LSTM layer
        self.lstm = nn.LSTM(input_size = embedding_dim, hidden_size = hidden_size, batch_first = True)
        # Layer 3 - Linear layer
        self.linear = nn.Linear(hidden_size, vocab_size)
    
    # initialize hidden states
    def init_hidden(self, batch_size, device):
        return (
            torch.zeros(1, batch_size, self.hidden_size).to(device),
            torch.zeros(1, batch_size, self.hidden_size).to(device)
        )
    
    # override forward function
    def forward(self, x, state = None, return_state = False):
        x = self.embedding(x)
        if state is None:
            state = self.init_hidden(x.size(0), x.device)
        out, state = self.lstm(x, state)
        out = self.linear(out)

        return out if not return_state else (out, state)
```

```python
vocab_size = len(vocab)
embedding_dim = 512
hidden_size = 1024
batch_size = 12

model = OurLSTMModel(embedding_dim, hidden_size, vocab_size)
model.to(device)

# print summary of model
print(model)
```

## 2.5 Training the model: loss and training operations
So now we have to train the model.
<br>
We can think of this problem as a standard classification problem. Given previous state of RNN and the input at given time step, predict the next character.
<br>
To train the model, we can use `Cross Entropy` loss. We will be computing the loss using true targets - `labels` and the predicted targets - `logits`.

```python
# Defining the loss function

# instantiate cross entropy loss
cross_entropy = nn.CrossEntropyLoss()

# define a function
def compute_loss(labels, logits):
    # labels: (batch_size, sequence_length)
    # logits: (batch_size, sequence_length, vocab_size)
    
    batched_labels = labels.view(-1) # flatten the matrix shape from (B,L) to (B*L,)
    
    # B,L,V = logits.shape
    batched_logits = logits.view(-1, logits.size(-1)) # reshape from (B,L,V) to (B*L,V)
    
    loss = cross_entropy(batched_logits, batched_labels)
    return loss
```

Let's start to train the model. But before that, it is important to define hyperparameters.

```python
# Set hyperparameters
vocab_size = len(vocab)

# Parameters
params = dict(
    num_training_iterations = 3000,
    batch_size = 8,
    seq_length = 100,
    learning_rate = 5e-3,
    embedding_dim = 256,
    hidden_size = 1024
)

# checkpoint locations
checkpoint_dir = './training_checkpoints'
checkpoint_prefix = os.path.join(checkpoint_dir, "my_ckpt")
os.makedirs(checkpoint_dir, exist_ok=True)
```

Before starting to train, we will first setup a Comet experiment to track training and model development.

```python
# Create comet expreiment
def create_experiment():
  # end any prior experiments
  if 'experiment' in locals():
    experiment.end()

  # initiate the comet experiment for tracking
  experiment = comet_ml.Experiment(
                  api_key=COMET_API_KEY,
                  project_name="6S191_Lab1_Part2")
  # log our hyperparameters, defined above, to the experiment
  for param, value in params.items():
    experiment.log_parameter(param, value)
  experiment.flush()

  return experiment
```

Now we will train the model.

```python
# define optimizer and training operation

# instantiate a model
model = OurLSTMModel(params['embedding_dim'], params['hidden_size'], vocab_size)

# move to GPU
model.to(device)

# instantiate optimizer
optimizer = torch.optim.Adam(model.parameters(), lr = params['learning_rate'])

def train_step(x,y):
    model.train()
    optimizer.zero_grad()

    # forward pass
    y_hat = model(x)
    # compute loss
    loss = compute_loss(y, y_hat)
    # backward pass
    loss.backward()
    optimizer.step()
    return loss

# Begin training
history = []
plotter = mdl.util.PeriodicPlotter(sec=2, xlabel='Iterations', ylabel='Loss')
experiment = create_experiment()
```

```python
if hasattr(tqdm, '_instances'): tqdm._instances.clear() # clear if it exists
for iter in tqdm(range(params["num_training_iterations"])):

    # Grab a batch and propagate it through the network
    x_batch, y_batch = create_batch(vectorized_songs, params["seq_length"], params["batch_size"])

    # Convert numpy arrays to PyTorch tensors
    x_batch = torch.tensor(x_batch, dtype=torch.long).to(device)
    y_batch = torch.tensor(y_batch, dtype=torch.long).to(device)

    # Take a train step
    loss = train_step(x_batch, y_batch)

    # Log the loss to the Comet interface
    experiment.log_metric("loss", loss.item(), step=iter)

    # Update the progress bar and visualize within notebook
    history.append(loss.item())
    plotter.plot(history)

    # Save model checkpoint
    if iter % 100 == 0:
        torch.save(model.state_dict(), checkpoint_prefix)

# Save the final trained model
torch.save(model.state_dict(), checkpoint_prefix)
experiment.flush()
```

## 2.6 Generate music using the trained RNN Model

```python
def generate_text(model, start_string, generation_length = 1000):
    input_idx = vectorize_string(start_string)
    input_idx = torch.tensor([input_idx], dtype = torch.long).to(device)

    # initialize hidden state
    state = model.init_hidden(input_idx.size(0), device)

    # empty string to store results
    text_generated = []
    tqdm._instances.clear()

    for i in tqdm(range(generation_length)):
        predictions, hidden_state = model(input_idx, state, return_state = True)
        predictions = predictions.squeeze(0)
        input_idx = torch.multinomial(torch.softmax(predictions, dim = -1), num_samples = 1)
        text_generated.append(idx2char[input_idx].item())
    return start_string + ''.join(text_generated)
```
